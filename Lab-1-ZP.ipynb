# Zadanie 1

```python
class Counter:
    def __init__(self):
        self.value = 0

    def plus_value(self):
        self.value += 1

    def minus_value(self):
        self.value -= 1
    
    def reset_counter(self):
        self.value = 0

    def get_value(self):
        return self.value
```

```python
# Program testujący

moj_licznik = Counter()

print(f'Wartość początkowa: {moj_licznik.get_value()}')

# Dodawanie wartości
moj_licznik.plus_value()
moj_licznik.plus_value()
print(f'Po dwóch dodawaniach: {moj_licznik.get_value()}')

# Odejmowanie wartości
moj_licznik.minus_value()
print(f'Po jednym odejmowaniu: {moj_licznik.get_value()}')

# Resetowanie licznika
moj_licznik.reset_counter()
print(f'Po zresetowaniu: {moj_licznik.get_value()}')
```

# Zadanie 2

```python
from dataclasses import dataclass

@dataclass
class Measurement:
    time: float
    value: float
    unit: str

p1 = Measurement(time=10.5, value=230.0, unit='V')
p2 = Measurement(time=12.0, value=245.5, unit='V')
p3 = Measurement(time=15.2, value=190.2, unit='V')
p4 = Measurement(time=18.1, value=213.7, unit='V')
```

```python
pomiary = [p1, p2, p3, p4]

print('Lista wszystkich pomiarów:')
for p in pomiary:
    print(p)

najwiekszy = max(pomiary, key=lambda p: p.value)

print(f'\nPomiar o najwyższej wartości to: {najwiekszy}')
```

# Zadanie 3

```python
import numpy as np

class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    # Wypisanie wektora
    def __str__(self):
        return f'[{self.x},{self.y}]'
    
    # Suma dwóch wektorów
    def __add__(self, other):
        return Vector2D(self.x + other.x, self.y + other.y)
    
    # Długość wektora
    def length(self):
        return np.sqrt(self.x**2 + self.y**2)

    # Mnożenie wektora przez skalar
    def __mul__(self, scalar):
        return Vector2D(self.x * scalar, self.y * scalar)
    def __rmul__(self, scalar):
        return Vector2D(self.x * scalar, self.y * scalar)
    
    # Iloczyn skalarny
    def dot_product(self, other):
        return float(self.x*other.x + self.y*other.y)
```

```python
v1 = Vector2D(3, 4)
v2 = Vector2D(1, 2)

print(f'Wektor 1: {v1}')
print(f'Wektor 2: {v2}')

v3 = v1 + v2
print(f'Suma v1 + v2: {v3}')

print(f'Długość v1: {v1.length()}')

v4 = v1 * 2
v5 = 10 * v1
print(f'v1 * 2: {v4}')
print(f'10 * v1: {v5}')

iloczyn = v1.dot_product(v2)
print(f'Iloczyn skalarny v1 i v2: {iloczyn}')
```

# Zadanie 4

```python
class MeasurementSeries:
    def __init__(self, measurements):
        self.measurements = list(measurements)

    # Pobieranie elementu przez indeks
    def __getitem__(self, index):
        return self.measurements[index]

    # Zmiana elementu przez indeks
    def __setitem__(self, index, value):
        self.measurements[index] = value

    # Zwrot liczby elementów przez len(obiekt)
    def __len__(self):
        return len(self.measurements)

    # Iteracja w pętli for
    def __iter__(self):
        return iter(self.measurements)

    # Liczenie średniej
    def average(self):
        return sum(self.measurements) / len(self.measurements)

    # Największy i najmniejszy pomiar
    def get_range(self):
        return min(self.measurements), max(self.measurements)
```

```python
data = [21.5, 22.0, 19.8, 25.3, 22.1]
series = MeasurementSeries(data)

print(f'Liczba pomiarów: {len(series)}')

print(f'Drugi pomiar: {series[1]}')

series[0] = 20.0
print(f'Pierwszy pomiar po zmianie: {series[0]}')

print('Wszystkie pomiary:')
for m in series:
    print(f"- {m}")

print(f'Średnia: {series.average():.2f}')
min, max = series.get_range()
print(f'Zakres: od {min} do {max}')
```

# Zadanie 5

```python
class Polynomial:
    def __init__(self, coefficients):
        self.coeffs = list(coefficients)

    def __repr__(self):
        terms = []
        for i, coeff in enumerate(self.coeffs):
            if coeff == 0: continue
            if i == 0:
                terms.append(f'{coeff}')
            elif i == 1:
                terms.append(f'{coeff}x')
            else:
                terms.append(f'{coeff}x^{i}')
        return ' + '.join(terms) if terms else '0'

    def __call__(self, x):
        result = 0
        for i, coeff in enumerate(self.coeffs):
            result += coeff * (x ** i)
        return result

    def degree(self):
        for i in range(len(self.coeffs) - 1, -1, -1):
            if self.coeffs[i] != 0:
                return i
        return 0

    def derivative_at(self, x):
        # Wzór na pochodną: (a*x^n)' = n * a * x^(n-1)
        derivative_value = 0
        for i in range(1, len(self.coeffs)):
            derivative_value += i * self.coeffs[i] * (x ** (i - 1))
        return derivative_value
```

```python
p = Polynomial([1, 0, 3])

print(f'Wielomian: {p}')
print(f'Stopień wielomianu: {p.degree()}')

x_val = 2
# 1 + 0*2 + 3*2^2 = 1 + 12 = 13
print(f'Wartość p({x_val}): {p(x_val)}')

# p'(x) = 0 + 0 + 6x
# p'(2) = 12
print(f'Wartość pochodnej p\'({x_val}): {p.derivative_at(x_val)}')
```

# Zadanie 6

```python
class Account:
    def __init__(self, username, points=0):
        self.username = username
        self.points = points

    # Dodawanie punktów do konta
    def add_points(self, amount):
        if amount > 0:
            self.points += amount
            print(f'Dodano {amount} pkt do konta {self.username}.')

    # Odejmowanie punktów z konta
    def subtract_points(self, amount):
        if amount > 0:
            self.points -= amount
            print(f'Odjęto {amount} pkt z konta {self.username}.')

    # Wypisanie stanu konta
    def display_info(self):
        print(f'Użytkownik: {self.username} | Punkty: {self.points}')


class PremiumAccount(Account):
    def __init__(self, username, points=0, bonus_multiplier=1.2):
        super().__init__(username, points)
        self.bonus_multiplier = bonus_multiplier

    # Punkty premium z bonusem
    def add_points(self, amount):
        bonus_amount = amount * self.bonus_multiplier
        print(f'Bonus x{self.bonus_multiplier}')
        super().add_points(bonus_amount)

class PenaltyAccount(Account):
    def __init__(self, username, points=0, fee=2):
        super().__init__(username, points)
        self.fee = fee

    # Dodawanie punktów z karą
    def add_points(self, amount):
        print(f'Pobrano opłatę: {self.fee} pkt.')
        super().add_points(amount - self.fee)

    # Odejmowanie punktów z karą
    def subtract_points(self, amount):
        print(f'Pobrano opłatę: {self.fee} pkt.')
        super().subtract_points(amount + self.fee)
```

```python
print('Konto Zwykłe')
acc1 = Account('Student PW', 100)
acc1.add_points(50)
acc1.display_info()

print('\nKonto Premium')
acc2 = PremiumAccount('Student Koźmińskiego', 100, bonus_multiplier=1.5)
acc2.add_points(50)
acc2.display_info()

print('\nKonto z Karą')
acc3 = PenaltyAccount('Student UW', 100, fee=5)
acc3.add_points(50)
acc3.display_info()
```

# Zadanie 7

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    def __init__(self, name, position=(0, 0)):
        self.name = name
        self.x, self.y = position

    @abstractmethod
    def sound(self):
        pass

    @abstractmethod
    def move(self):
        pass

    def info(self):
        print(f'[{self.__class__.__name__}] {self.name} jest na pozycji ({self.x}, {self.y})')


class Mrowka(Animal):
    def __init__(self, name, position=(0, 0)):
        super().__init__(name, position)

    def sound(self):
        return 'Mrówka milczy, ale dużo pracuje!'

    def move(self):
        self.x += 2
        print(f'{self.name} tupie po ziemi.')


class Wrobel(Animal):
    def __init__(self, name, position=(0, 0)):
        super().__init__(name, position)

    def sound(self):
        return 'Cwir! Cwir!'

    def move(self):
        self.x += 10
        self.y += 10
        print(f'{self.name} leci wysoko!')

class Wilk(Animal):
    def __init__(self, name, position=(0, 0)):
        super().__init__(name, position)

    def sound(self):
        return 'Auuuuuuu!'

    def move(self):
        self.x += 5
        self.y += 0
        print(f'{self.name} biegnie szybko!')
```

```python
zoo = [
    Mrowka('Maciek', (0,0)),
    Wrobel('Wojtek', (10,10)),
    Wilk('Michał', (2,0))
]

for animal in zoo:
    animal.info()
    print(f'Dźwięk: {animal.sound()}')
    animal.move()
    animal.info()
    print('\n')
```
